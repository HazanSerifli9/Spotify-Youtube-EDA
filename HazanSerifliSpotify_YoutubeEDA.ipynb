```python
# Identifying Categorical Columns
cat_cols = [col for col in music.columns if str(music[col].dtypes) in ["category", "object", "bool"]]

# Identifying Numeric Columns with Few Unique Values (Potentially Categorical)
num_but_cat = [col for col in music.columns if music[col].nunique() < 10 and str(music[col].dtypes) in ["int", "float"]]

# Identifying Categorical Columns with Many Unique Values
cat_but_car = [col for col in music.columns if music[col].nunique() > 20 and str(music[col].dtypes) in ["category", "object"]]

# Combining Categorical Columns
cat_cols = cat_cols + num_but_cat

# Removing Columns with Too Many Unique Values
cat_cols = [col for col in cat_cols if col not in cat_but_car]

# Defining and Using a Summary Function
def cat_summary(musicframe, col_name, plot=False, pie=False):
    summary_df = pd.DataFrame({
        col_name: musicframe[col_name].value_counts(),
        "Ratio (%)": 100 * musicframe[col_name].value_counts() / len(musicframe)
    })
    
    print(summary_df)
    print("####################################################################")
    
    # Define a consistent color palette
    palette = sns.color_palette("Set2")  # You can change "Set2" to any other Seaborn or custom palette

    if plot:
        plt.figure(figsize=(10, 5))
        sns.countplot(x=col_name, data=musicframe, 
                      order=musicframe[col_name].value_counts().index, 
                      palette=palette)
        plt.title(f'Distribution of {col_name}')
        plt.xticks(rotation=45)
        plt.show()

    if pie:
        plt.figure(figsize=(8, 8))
        size = musicframe[col_name].value_counts()
        colors = sns.color_palette("Set2", len(size))  # Ensure pie chart uses the same palette
        plt.pie(size, labels=size.index, autopct='%1.1f%%', startangle=140, colors=colors)
        plt.title(f'Pie Chart of {col_name}')
        plt.show()

# Apply the summary function to each categorical column
for col in cat_cols:
    cat_summary(music, col, plot=True, pie=True)
```

```python


# Define a consistent color palette
palette = sns.color_palette("Set2")

# Select only numeric columns
num_cols = music.select_dtypes(include=['int64', 'float64']).columns

# Defining the summary function for numerical variables
def num_summary(musicframe, numerical_col, plot=False, kde=False):
    quantiles = [0.05, 0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 0.95, 0.99]
    summary_df = musicframe[numerical_col].describe(quantiles).T
    print(summary_df)
    print("####################################################################")

    if plot:
        plt.figure(figsize=(10, 5))
        sns.histplot(musicframe[numerical_col], bins=30, kde=True, color=palette[0])  # Use first color from palette
        plt.title(f'Distribution of {numerical_col}')
        plt.show()
    if kde:
        plt.figure(figsize=(10, 5))
        sns.kdeplot(musicframe[numerical_col], fill=True, color=palette[1])  # Use second color from palette
        plt.title(f'KDE Plot of {numerical_col}')
        plt.show()

# Applying the function to each numerical column
for col in num_cols:
    num_summary(music, col, plot=True, kde=True)
```

```python


# Select only numeric columns
numeric_cols = music.select_dtypes(include=[np.number])

# Calculate the correlation matrix
correlation = numeric_cols.corr()

# Define the colormap using the consistent palette
cmap = sns.color_palette("coolwarm", as_cmap=True)  # Change "coolwarm" to align visually with Set2 style

def magnify():
    return [dict(selector="th",
                 props=[("font-size", "7pt")]),
            dict(selector="td",
                 props=[('padding', "0em 0em")]),
            dict(selector="th:hover",
                 props=[("font-size", "12pt")]),
            dict(selector="tr:hover td:hover",
                 props=[('max-width', '200px'),
                        ('font-size', '12pt')])
]

# Styled correlation table
correlation.style.background_gradient(cmap, axis=1)\
    .format(precision=3)\
    .set_properties(**{'max-width': '80px', 'font-size': '10pt'})\
    .set_caption("Hover to magnify")\
    .set_table_styles(magnify())

# Create a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap=cmap, vmin=-1, vmax=1, 
            cbar_kws={"shrink": 0.8}, linewidths=0.5)
plt.title('Correlation Heatmap between Numeric Features')
plt.show()
```

# 3. Visualization

```python
#distrubtion of track durations

# Assuming the correct column name is 'Duration_ms' and converting it to minutes
music['Duration_Minutes'] = music['Duration_ms'] / 60000

fig, ax = plt.subplots(figsize=(10, 6)) 
sns.histplot(data=music, x="Duration_Minutes", ax=ax, kde=True, color=palette[0])  
ax.set(xlabel='Song Duration in Minutes', 
       ylabel='Number of Songs', 
       title='Distribution of Song Duration by Minutes', 
       xlim=(0, 10))  
plt.tight_layout()
plt.show()
```

Generally, the song duration is between 3 to 4 minutes.

```python
# Assuming the consistent color palette is defined
palette = sns.color_palette("Set2")

# Creating the box plot for song durations
fig, ax = plt.subplots(figsize=(10, 6))
sns.boxplot(data=music, x="Duration_Minutes", ax=ax, color=palette[1])  # Use the second color from the palette
ax.set(
    xlabel='Song Duration in Minutes',
    title='Box Plot of Song Duration by Minutes'
)
plt.tight_layout()
plt.show()
```

```python
 #Create custom groups based on duration in minutes
def categorize_duration(duration):
    if duration < 3:
        return 'Short'
    elif 3 <= duration < 5:
        return 'Medium'
    else:
        return 'Long'

music['duration_group'] = music['Duration_Minutes'].apply(categorize_duration)

# Visualize the groups using a box plot
plt.figure(figsize=(12, 6))
sns.boxplot(x='duration_group', y='Duration_Minutes', data=music, palette=palette)
plt.title('Box Plot of Song Duration by Groups')
plt.xlabel('Duration Group')
plt.ylabel('Duration (minutes)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python


# Create scatter plots with regression lines
plt.figure(figsize=(14, 6))

# Scatter plot for Views vs Duration
plt.subplot(1, 2, 1)
sns.regplot(x='Duration_Minutes', y='Views', data=music, scatter_kws={'s':10}, line_kws={'color':'red'})
plt.title('Views vs Duration')
plt.xlabel('Duration (minutes)')
plt.ylabel('Views')
plt.grid(True, linestyle='--', alpha=0.7)

# Scatter plot for Likes vs Duration
plt.subplot(1, 2, 2)
sns.regplot(x='Duration_Minutes', y='Likes', data=music, scatter_kws={'s':10}, line_kws={'color':'blue'})
plt.title('Likes vs Duration')
plt.xlabel('Duration (minutes)')
plt.ylabel('Likes')
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

The relationship between likes and views is weak. However, we can see that people prefer short videos. Most views and likes are concentrated on short-duration videos. This suggests that viewers generally show more interest in shorter content.

```python
music['duration_group'] = music['Duration_Minutes'].apply(categorize_duration)

# Aggregate views for each duration group
views_by_group = music.groupby('duration_group')['Views'].sum()

# Create a pie chart
plt.figure(figsize=(8, 8))
views_by_group.plot.pie(autopct='%1.1f%%', colors=['#ff9999','#66b3ff','#99ff99'])
plt.title('Distribution of Views by Duration Group')
plt.ylabel('')  # Hide the y-label
plt.show()
```

Medium and short videos have higher views than long videos. People are more interested in medium and short videos.

```python
from sklearn.preprocessing import StandardScaler
import plotly.graph_objects as go

num_cols = music.select_dtypes(include=['int64', 'float64']).columns

num_cols = num_cols.drop(['Stream', 'Comments', 'Likes', 'Views'])


scaler = StandardScaler()


df_scaled = pd.DataFrame(scaler.fit_transform(music[num_cols]), columns=num_cols)


# Create a 'Popularity_Index' column based on 'Views'
music['Popularity_Index'] = pd.qcut(music['Views'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])

df_scaled['Popularity_Index'] = music['Popularity_Index']


traces = []

for cat in df_scaled['Popularity_Index'].unique():
    data = df_scaled[df_scaled['Popularity_Index'] == cat][num_cols].mean().tolist()
    trace = go.Scatterpolar(
        r=data + data[:1],
        theta=num_cols.tolist() + [num_cols[0]],
        fill='toself',
        name=cat
    )
    traces.append(trace)

# Create the layout
layout = go.Layout(
    polar=dict(
        radialaxis=dict(visible=True),
    ),
    showlegend=True,
    title='Features Radar Plot by Popularity Group '
)

# Create the figure
fig = go.Figure(data=traces, layout=layout)

# Show the figure
fig.show()
```

When we compare popularity group we can see that songs in the "Very High" popularity group may have higher Loudness and Energy, while "Low" popularity songs might show higher Acousticness or Instrumentalness.

```python
# Assuming the consistent color palette is defined
palette = sns.color_palette("Set2", n_colors=10)  # Adjusted to match the number of colors if needed

for group in music['Popularity_Index'].unique():
    data = music[(music['Popularity_Index'] == group) & (music['Album_type'] == 'single')].copy()
    data['TArtist'] = data.groupby('Album')['Artist'].transform(lambda x: ', '.join(x.unique()))
    data = data.groupby('TArtist').Stream.max().sort_values(ascending=False).head(5).reset_index()

    if not data.empty:
        plt.figure(figsize=(10, 5))
        ax = sns.barplot(x='Stream', y='TArtist', data=data, palette=palette)  # Use the consistent palette
        plt.title(f'Top 5 Artists on Spotify - {group}')
        plt.xlabel('Streams')
        plt.ylabel('Artists')
        plt.tight_layout()
        plt.show()
```

```python
popular_songs = music.groupby('Track').agg({'Views': 'sum', 'Likes': 'sum', 'Comments': 'mean'}).reset_index()
popular_songs = popular_songs.sort_values(by='Views', ascending=False).head(10)

# Aggregate data to find the most popular Artists
popular_Artists = music.groupby('Artist').agg({'Views': 'sum', 'Likes': 'sum', 'Comments': 'mean'}).reset_index()
popular_Artists = popular_Artists.sort_values(by='Views', ascending=False).head(10)

# Visualize the most popular songs
plt.figure(figsize=(12, 8))
sns.barplot(x='Views', y='Track', data=popular_songs, palette=palette)
plt.title('Top 10 Most Popular Songs by Play Count')
plt.xlabel('Play Count')
plt.ylabel('Song')
plt.show()

# Visualize the most popular Artists
plt.figure(figsize=(12, 8))
sns.barplot(x='Views', y='Artist', data=popular_Artists, palette=palette)
plt.title('Top 10 Most Popular Artists by Play Count')
plt.xlabel('Play Count')
plt.ylabel('Artist')
plt.show()
```

```python
# Assuming the consistent color palette is defined
palette = sns.color_palette("Set2", n_colors=10)  # Generate 10 colors for the top 10 artists

# Prepare the data
top_artists = music.groupby('Artist')[['Likes', 'Comments']].sum().sort_values('Likes', ascending=False).head(10)

# Plot the data
ax = top_artists.plot(
    kind='bar',
    figsize=(10, 6),
    color=[palette[0], palette[1]],  # Use two colors from the palette for Likes and Comments
    edgecolor='black'  # Optional: add edges for a cleaner look
)
ax.set_title('Top 10 Artists by Likes and Comments', fontsize=14)
ax.set_xlabel('Artist', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()
```
